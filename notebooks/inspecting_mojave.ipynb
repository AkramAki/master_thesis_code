```python
from pathlib import Path

import torch
from torchvision.utils import make_grid

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import PowerNorm
from PIL import Image
from IPython.display import display

from radio_diffusion.data.mojave import MojaveDataModule
```

```python
img_size=256
image_dir = Path(f"/cephfs/users/aaki/data/Mojave_I_{img_size}")

data = MojaveDataModule(
    data_dir=image_dir,
    batch_size=16,
    num_workers=0,
    val_fraction=0.1,
    test_fraction=0.1,
    seed=42,
)
```

```python
data.setup("fit")
train_loader = data.train_dataloader()
images = next(iter(train_loader))

print(f"Training images: {len(data.train_dataset)}")
print(f"Validation images: {len(data.val_dataset)}")
print(f"Batch shape: {tuple(images.shape)}; dtype: {images.dtype}")
print(f"Pixel range: [{images.min().item():.3f}, {images.max().item():.3f}]")
print(f"Image shape: {images[0].shape}")
plt.imshow(images[0,0])
plt.colorbar()
None
```

```python
grid = make_grid(images, nrow=4)
preview = grid[0].cpu().numpy()  # Grayscale channel

# Power-law normalization
norm = PowerNorm(gamma=0.5, vmin=preview.min(), vmax=preview.max())

fig, ax = plt.subplots(figsize=(12, 8))

im = ax.imshow(preview, cmap="inferno", norm=norm)
fig.colorbar(im, ax=ax, label="Pixel intensity")

ax.axis("off")
None
```
